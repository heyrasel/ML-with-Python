```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# For data entry
x = np.arange(11) # 0 to 10 porjonto data
y = x ** 2 # y = x^2
print(x, y)
```

```python
# for scatter graph plotting
plt.scatter(x,y) # x axis e x and y axis e y bosbe
```

```python
# line diye korte chaile
plt.plot(x,y, 'r-*') # line er color red hobe for r and point gula * diye represent korbe.
```

```python
# for title in graph
plt.plot(x,y, 'r-*')
plt.xlabel('X axis lebel')
plt.ylabel('Y axis label')
plt.title('Figure title')
plt.xticks(range(0,11)) # x-axis er range o theke 10 porjonto. 1 1 kore barbe.
plt.yticks(range(0,101,5)) #y-axis er range o theke 10 porjonto. 5 kore barbe.
plt.show() # for removing garbage value
```

```python
# Subplotting mane onekgula plotting.
plt.subplot(3,3,1) # first number ta-> 3 mane 1/3 vag e dekhabe row ta. 2nd number mean 3ta col, 3 number mean graph tar position.
plt.plot(x,y,'r-*')

plt.subplot(1,3,2)
plt.plot(x,y,'g-*')

plt.subplot(1,3,3)
plt.plot(x,y,'b--')
```

```python
# Plotting figure and save as pdf
fig = plt.figure(figsize=(10,5)) # 10 holo width and 5 is height.

axes = fig.add_axes([0.1,0.2,0.5,0.3]) # jei pdf e save hobe sei pdf left=10% side e, bottom=20%, width=50%, height=30% hobe.
axes.plot(x,y,'r-*')
axes.set_xlabel('x label')
axes.set_ylabel('y label')
axes.set_title('Test Figure')

fig.savefig('Graph1.pdf') # jpeg, png etc format eo save kora jabe.
```

```python
# Subplotting, ekta graph er moddhe onek graph
fig, axes = plt.subplots(figsize=(10,8))

axes.plot(x, x+1, color="red", linewidth=0.25 ) # y ta holo x+1, jei line ta asbe tar width.
axes.plot(x, x+2, color="red", linewidth=0.50 )
axes.plot(x, x+3, color="red", linewidth=1.00 )
axes.plot(x, x+4, color="red", linewidth=2.00 )

axes.plot(x, x+5, color="green", lw=0.50, linestyle = '-' ) # linewidth ke lw likha jai. linestyle ke ls.
axes.plot(x, x+6, color="green", lw=1.00, ls='-.' )
axes.plot(x, x+7, color="green", lw=2.00, ls=':' ) # : deya ta dot dot kore asche.

axes.plot(x, x+5, color="blue", lw=3, linestyle = '--', marker='o', markersize=8 ) # marker mane point gula. o mane circle hobe.
axes.plot(x, x+6, color="blue", lw=3, ls='-.', marker='s', markersize=8, markerfacecolor="yellow", 
         markeredgecolor="green", markeredgewidth=3) # markerface yello mane vetore yellow and edge e green colour. 
axes.plot(x, x+7, color="blue", lw=3, ls=':', marker='1', markersize=12 ) # marker=1 eta emn > size hoy.
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 📊 Step 1: Mock Data Creation

np.random.seed(42)  # reproducibility

districts = ['Downtown', 'East Side', 'West End', 'Northville', 'Southtown']
complaint_types = ['Power Cut', 'Trafic', 'Traffick', 'Noise', 'Garbage', 'Water Leak', 'Power Cut', 'Ice Cream Cart']
ice_cream_flavors = ['Vanilla', 'Chocolate', 'Strawberry', 'Mint', 'Mango', 'Lemon', 'Pistachio', 'Cookie Dough', 'Coffee', 'Blueberry']

n_rows = 500

data = {
    'district': np.random.choice(districts, size=n_rows), # district theke randomly 500 ta district name generate korbe.
    'complaint_type': np.random.choice(complaint_types, size=n_rows), # same
    'temperature': np.random.normal(loc=32, scale=5, size=n_rows), # loc=32 mane middle point 32 and scale=5 mane middle thek +-5 porjonto data generate hobe(0 to 5)
    'electricity_usage': np.random.normal(loc=250, scale=50, size=n_rows),  # same
    'num_ice_cream_carts': np.random.choice([0, 1, 2, 5, 10, 10, 10], size=n_rows), # 10 besi deya jate biasness create hoy
    'survey_rating': np.random.choice([1, 2, 3, 4, 5, np.nan], size=n_rows), # np.nan means some people doesnt take the survey
}
df = pd.DataFrame(data)
df
```

```python
# Creating some duplicate value for preprocessing
df.loc[5:10, 'complaint_type']= np.nan # c_type er 5 theke 10 row porjonto null value dilam.
df = pd.concat([df, df.iloc[0:5]]) # 1st 5 ta row abar sesh e dhukai dicchi.
df
```

```python
# Dataset er ekta histogram plot korbo
sns.set(style='whitegrid') # background white and grid thakbe
plt.figure(figsize=(8,4)) # height*width
sns.histplot(df['temperature'], kde=True, bins=10) # kde mane line ta. bins mane x axis e koyta segment korbe.
plt.title("Histogram of Temperature")
plt.xlabel('temp in degree')
plt.ylabel('frequency')
plt.show()
```
