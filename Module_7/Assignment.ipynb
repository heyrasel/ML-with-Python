# Scenario
An e-commerce store recorded the number of items bought by five recent customers: [1, 2, 2, 3, 42]. The analytics team also found a correlation of r = 0.88 between "minutes spent on the website" and "total money spent." Additionally, the daily website traffic follows a standard normal distribution.

Questions :

1. Is the "number of items bought" considered qualitative or quantitative data?

2. Calculate the mean, median, and mode for the five customer orders.

3. Due to the outlier (42) in the data, which measure of central tendency gives the most accurate representation of a typical customer, and why?

4. Interpret the correlation of r = 0.88. Does this prove that staying on the site longer causes people to spend more money?

5. According to the Empirical Rule, what percentage of daily website traffic falls within exactly one standard deviation of the mean?



```python
Ans of Q1: This is quantitative data.

Ans of Q2: Mean = (1+2+2+3+42)/5 = 10. Median = 2. Mode = 2

Ans of Q3: The most accurate measure is Median. Because, as the last data(42) is extreme outlier so it affects median too much. So Median is the accurate representation for this case.

Ans of Q4: As r = 0.88 so it defines strong positive correlation. It means as the time on the website increases money spend tends to increse. But it does not prove that spending more time causes higher spending. It also depends on people interest product type, quality etc.

Ans of Q5: Using the Empirical Rule(68–95–99.7) 68% of data falls within +-1 standard deviation.
```
