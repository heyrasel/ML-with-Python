```python
import pandas as pd
import numpy as np

df = pd.DataFrame(
    {
        'A': [21, np.nan, 22, 4, np.nan],
        'B': [31, 32, 33, 34, 35],
        'C': [np.nan, np.nan, np.nan, np.nan, np.nan],
        'D': [46, 47, np.nan, 48, 49]
    }
)
display(df)

# For count total NaN values
def count_nan(df):
    return df.isna().sum().sum()

print(f"Total NaN: {count_nan(df)}")

# Forward fill
df_ffill = df.ffill()
display(df_ffill)

# Backwad fill
df_bfill = df.bfill()
display(df_bfill)

# Fill with 0
df_zero = df.fillna(0)
display(df_zero)
```
